```python
import numpy as np
import pandas as pd
#load datasetn
df= pd.read_csv("/titanic3.csv")
#extract specific columns as Numpy arrays
ages = df['age'].values
fares = df['fare'].values
survived = df['survived'].values
#statistical analysis
mean_fare = np.nanmean(fares)
median_fare = np.nanmedian(fares)
std_age = np.nanstd(ages) #nanstd ignores missing NaN values
youngest = np.nanmin(ages)
highest_fare = np.nanmax(fares)
#filter fares of survived vs deceased passengers
survived_mask= (survived == 1)
mean_fare_survived = np.nanmean(fares[survived_mask])
mean_fare_deceased = np.nanmean(fares[~survived_mask])
pclass = df['pclass'].values
child_first_class_mask=(ages<18)&(pclass==1)
total_count=np.sum(child_first_class_mask)
#determine global survival rate(Survived column contains 0s/1s)
survival_rate = np.mean(survived)*100
```

```python
print(f"Mean Fare : {mean_fare}")
print(f"Max fare: {highest_fare}")
print(f"Total count: {total_count}")
print(f"Mean fare deceased: {mean_fare_deceased}")
print(f"Mean fare survived: {mean_fare_survived}")
print(f"Global survival rate: {survival_rate}%")
```
