```python
import pandas as pd
import seaborn as sns
import missingno as mn
import matplotlib.pyplot as plt
     
```

```python
df_titanic=pd.read_csv(r"Y:\New folder\titanic3.csv")
```

```python
display(df_titanic.age.isnull())
```

```python
print(df_titanic.isnull().mean()*100)
```

```python
mean_age=df_titanic['age'].mean()
```

```python
print(mean_age)
```

```python
df_titanic['age_mean']=df_titanic['age'].fillna(mean_age)
```

```python
median_age=df_titanic['age'].median()
```

```python
print(median_age)
```

```python
df_titanic['age_median']=df_titanic['age'].fillna(median_age)
```

```python
fig=plt.figure()
```

```python
ax=fig.add_subplot(111)
```

```python
df_titanic['age'].plot(kind='kde',ax=ax)
```

```python
df_titanic ['age_median'].plot(kind='kde',ax=ax,color='red')
```

```python
df_titanic['age_mean'].plot(kind='kde',ax=ax,color='green')
```
