```python
from sklearn.datasets import fetch_20newsgroups
newsgroups = fetch_20newsgroups(subset="test")
data = newsgroups.data
```

```python
import pandas as pd
df= pd.DataFrame(data,columns=["text"])
print(df.describe())
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import matplotlib.pyplot as plt
vectorizer = CountVectorizer(stop_words="english",max_features=5000)
X_counts = vectorizer.fit_transform(data)
```

```python
word_counts = X_counts.toarray().sum(axis=0)
word_freq = [(word,word_counts[idx])for word, idx in vectorizer.vocabulary_.items()]
```

```python
word_freq = sorted(word_freq, key=lambda x: x[1], reverse=True)
words = [wf[0] for wf in word_freq[:50]]
counts = [wf[1] for wf in word_freq[:50]]
```

```python
plt.figure(figsize=(10,5))
plt.bar(words,counts)
plt.xticks(rotation=90)
plt.title("Top 50 word frequency")
plt.show()
```
