```python
from sklearn.datasets import fetch_california_housing
```

```python
california_housing= fetch_california_housing(as_frame=True)
```

```python
import matplotlib.pyplot as plt
```

```python
california_housing.frame.hist(figsize=(17, 14), bins=30, edgecolor="black")
plt.subplots_adjust(hspace=0.7,wspace=0.4)
```

```python
plt.subplots_adjust(hspace=0.7, wspace=0.4)

features_of_interest = ["AveRooms", "AveBedrooms", "AveOccup", "Population"]
california_housing.frame[features_of_interest].describe()
```

```python
import seaborn as sns
```

```python
sns.scatterplot(data=california_housing.frame,x="Longitude",y="Latitude",size="MedHouseVal",hue="MedHouseVal",palette="viridis",alpha=0.5)
plt.legend(title="MedHouseVal", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.title("Median house values depending on their spatial location")
```
