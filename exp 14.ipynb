```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#1.sample dataset: monthly closing prices (in usd)
data={ 'Date':['2025-01','2025-02','2025-03','2025-04','2025-05','2025-06'],
       'Tech_ETF': [410.0, 425.5, 418.0, 440.2, 455.0, 470.5],
       'Broad_Market':[500.0, 512.0, 508.5, 524.0, 535.0, 548.0],
       'Treasury_Bonds': [95.0, 93.5, 96.0, 92.0, 91.5, 90.0],
       'Gold_ETF':[190.0, 192.0, 198.5, 201.0, 199.0, 205.0]
     }
```

```python
df = pd.DataFrame(data).set_index('Date')
```

```python
#2. calsulate monthly percentage returns and correlation matrix
returns=df.pct_change().dropna()
corr_matrix = returns.corr()
```

```python
#3. plot the heat map
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix, #show correlation numbers inside cells
    annot=True, #Round to 2 decimal places
    cmap="RdYlGn", #Red (negative)-yellow (neutral)-Greem(positive)
    vmin=-1, vmax=1, #Anchor color scale from -1 to +1
    linewidths=0.5
)
```
