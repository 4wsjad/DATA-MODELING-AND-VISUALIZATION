```python
import pandas as pd
import matplotlib.pyplot as plt
df_titanic = pd.read_csv(r"/content/drive/MyDrive/Colab Notebooks/titanic3.csv")
```

```python
survival_rates=df_titanic.groupby(['pclass','sex'])['survived'].mean().unstack()
display(survival_rates)
survival_rates.plot(kind='bar',figsize=(10,6))
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.xticks(rotation=0)
plt.legend(title='sex')
plt.grid(axis='y',linestyle='--',alpha=0.7)
plt.tight_layout()
plt.show()
```
