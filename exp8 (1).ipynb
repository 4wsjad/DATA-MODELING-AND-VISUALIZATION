```python
from scipy import stats
```

```python
df = pd.read_csv(r"C:\Users\20251BCD0012\Downloads\titanic3.csv")
```

```python
age_non_survivors = df[df['survived'] == 0]['age'].dropna()
```

```python
age_survivors = df[df['survived'] == 1]['age'].dropna()
```

```python
t_stat, p_val = stats.ttest_ind(age_non_survivors,age_survivors,equal_var=False)
```

```python
t_stat, p_val
```

```python
contingency_table =pd.crosstab(df['survived'],df['pclass'])
```

```python
chi2 , p_val, dof, expected = stats.chi2_contingency(contingency_table)
```
