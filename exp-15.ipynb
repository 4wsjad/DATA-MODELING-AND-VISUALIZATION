```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use("Agg")
```

```python
print("==== Program 15: Visualization on streaming datasets(Stock market & weather)====\n");
np.random.seed(8)
```

```python
def simulate_stream(n_points, start_val, vol):
    stream = [start_val]
    for _ in range(n_points - 1):
        stream.append(stream[-1] + np.random.normal(0, vol))
    return stream
```

```python
stock_stream = simulate_stream(100, 100, 2)
temp_stream = simulate_stream(100, 25, 1)
```

```python
fig, axes = plt.subplots(1,2, figsize=(13, 4.5))
```

```python
axes[0].plot(stock_stream, color="green")
axes[0].set_title("streaming " "stock price Ticks (Live Feed Simulation)")
axes[0].set_xlabel("Tick #")
axes[0].set_ylabel("Price")
```

```python
axes[1].plot(temp_stream, color="orangered")
axes[1].set_title("Streaming Weather " "Sensor Data (Live Feed Simulation)")
axes[1].set_xlabel("Reading #")
axes[1].set_ylabel("Temperature (C)")
```
